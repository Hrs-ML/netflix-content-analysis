### Import dataset

```python
import pandas as pd
```

```python
df = pd.read_csv(r'E:\internship\Auspify\netflix_dataset.csv')
```

```python
df.info()
```

#### Above Result shows that there is no null or missing value.

### Correct Data Type

```python
df["date_added"] = pd.to_datetime(df["date_added"].str.strip()) ## Chnaged dtype str to datetime
```

```python
# Extract the numbers and convert into integers
df["duration"] = df["duration"].str.extract(r'(\d+)').astype('int64')
```

### Check duplicates

```python
df.duplicated().sum()
```

##### No duplicates found

### Check Formating Inconsistencies

```python
print(df["type"].value_counts())
print()  
print(df["rating"].value_counts())
```

```python
print(df["country"].nunique())
```

```python
# Check country formatting
country_clean = df["country"].str.strip()

print("Extra spaces:", (df["country"] != country_clean).sum())

# Check case differences
print("Original unique:", df["country"].nunique())
print("Lowercase unique:", df["country"].str.lower().nunique())
```

##### There is no formatting inconsistenc.

### Standardize Columns

```python
# Standardize Rating
df["rating"] = df["rating"].str.strip().str.upper()

# Standardize Type
df["type"] = df["type"].str.strip().str.title()
```

```python
df.head()
```

### Save dataset as a CSV file

```python
df.to_csv("cleaned_netflix_dataset.csv", index=False)

print(f"Export complete! Successfully saved {len(df)} rows to 'cleaned_netflix_dataset.csv'.")
```
