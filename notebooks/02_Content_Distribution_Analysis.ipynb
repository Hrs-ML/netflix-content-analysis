### Import Dataset

```python
import pandas as pd
df = pd.read_csv(f'E:\internship\Auspify\cleaned_netflix_dataset.csv')
df.head()
```

### Total Number of Movies and TV Shows

```python
movie_count = (df['type'] == 'Movie').sum()
tv_show_count = (df['type'] == 'Tv Show').sum()
total = movie_count + tv_show_count

print(f'Total Movies: {movie_count}')
print(f'Total TV Shows: {tv_show_count}')
print(f'Combined Total: {total}')
```

### Visualizations for Content Distribution.

```python
import matplotlib.pyplot as plt

# Count Movies and TV Shows
content_counts = df["type"].value_counts()

# Create two charts side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

fig.suptitle("Distribution of Movies and TV Shows", fontsize=18, fontweight="bold")

# Bar Chart
bars = axes[0].bar(content_counts.index, content_counts.values)
axes[0].set_title("Total Number of Movies and TV Shows", pad=15, fontweight="bold")
axes[0].set_xlabel("Content Type")
axes[0].set_ylabel("Number of Titles")
axes[0].spines["top"].set_visible(False)
axes[0].spines["right"].set_visible(False)
axes[0].bar_label(bars, fmt="%d")

# Pie Chart
axes[1].pie(content_counts.values,labels=content_counts.index,autopct="%1.1f%%")
axes[1].set_title("Proportion of Movies and TV Shows", pad=15, fontweight="bold")

# Display both charts
plt.tight_layout()
plt.show()
```

### Content Proportions Comparison

```python
# Calculate the proportion of each content type
content_proportion = (content_counts / content_counts.sum()) * 100

# Display the proportion of each content type
print("Content Proportions:")
for content_type, percentage in content_proportion.items():
    print(f"{content_type}: {percentage:.2f}%")

# Calculate the difference between Movies and TV Shows
difference = (
    content_proportion["Movie"] -
    content_proportion["Tv Show"]
)

print(f"\nDifference: {difference:.2f} percentage points")

# Compare which content type has the larger proportion
if content_proportion["Movie"] > content_proportion["Tv Show"]:
    print("Movies have a higher proportion than TV Shows.")
else:
    print("TV Shows have a higher proportion than Movies.")
```


## Key Findings

- Movies are the majority of the content on Netflix, with around 70% of the total titles.
- TV Shows make up around 30% of the total content.
- Movies have a significantly higher proportion than TV Shows, with a difference of around 39 percentage points.
- The analysis shows that Netflix's content library is more focused on Movies than TV Shows.
